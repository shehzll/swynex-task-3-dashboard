```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
```

```python

print("TOTAL ROWS:", len(df))
print("TOTAL COLUMNS:", len(df.columns))

print("\n--- NUMERICAL STATISTICS ---")
display(df.describe())

print("\n--- MISSING VALUES ---")
display(df.isnull().sum())

print("\n--- DUPLICATE ROWS ---")
print(df.duplicated().sum())

print("\n--- DATA TYPES ---")
print(df.dtypes)
```

```python

counts = df["Survived"].value_counts().reindex([0, 1], fill_value=0)

plt.figure(figsize=(7, 4))
plt.bar(["Did not survive", "Survived"], counts.values)

plt.title("Titanic Passenger Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()

print("Survival counts:")
print(counts.rename(index={0: "Did not survive", 1: "Survived"}))
```

```python

plt.figure(figsize=(7, 4))

plt.hist(df["Age"].dropna(), bins=20, edgecolor="black")

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.show()
```

```python

avg_fare = df.groupby("Survived")["Fare"].mean().reindex([0, 1])

plt.figure(figsize=(7, 4))

plt.bar(["Did not survive", "Survived"], avg_fare.values)

plt.title("Average Ticket Fare by Survival Status")
plt.xlabel("Survival Status")
plt.ylabel("Average Fare")

plt.tight_layout()
plt.show()

print("Average fare by survival status:")
print(avg_fare.rename(index={0: "Did not survive", 1: "Survived"}).round(2))
```

```python

survival_by_sex = df.groupby("Sex_male")["Survived"].mean() * 100

labels = [
    "Female" if value == 0 else "Male"
    for value in survival_by_sex.index
]

plt.figure(figsize=(7, 4))

plt.bar(labels, survival_by_sex.values)

plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)

plt.tight_layout()
plt.show()

print("Survival rate by sex (%):")
print(
    survival_by_sex.rename(index={0: "Female", 1: "Male"}).round(2)
)
```

```python

# Insight 1: Overall survival rate
print("INSIGHT 1: Overall Survival Rate")
print(round(df["Survived"].mean() * 100, 2), "%")

# Insight 2: Survival counts
print("\nINSIGHT 2: Survival Counts")
print(
    df["Survived"]
    .value_counts()
    .rename(index={0: "Did not survive", 1: "Survived"})
)

# Insight 3: Survival rate by sex
print("\nINSIGHT 3: Survival Rate by Sex (%)")
print(
    (df.groupby("Sex_male")["Survived"].mean() * 100)
    .rename(index={0: "Female", 1: "Male"})
    .round(2)
)

# Insight 4: Average fare by survival status
print("\nINSIGHT 4: Average Fare by Survival Status")
print(
    df.groupby("Survived")["Fare"]
    .mean()
    .rename(index={0: "Did not survive", 1: "Survived"})
    .round(2)
)

# Insight 5: Survival rate by age group
age_bins = [0, 12, 18, 35, 60, 120]
age_labels = [
    "Child", "Teenager", "Young Adult", "Adult", "Senior"
]

age_group = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)

print("\nINSIGHT 5: Survival Rate by Age Group (%)")
print(
    df.groupby(age_group, observed=False)["Survived"]
    .mean()
    .mul(100)
    .round(2)
)
```

```python

print("--- AGE RANGE ---")
print("Minimum age:", df["Age"].min())
print("Maximum age:", df["Age"].max())

print("\n--- FARE RANGE ---")
print("Minimum fare:", df["Fare"].min())
print("Maximum fare:", df["Fare"].max())

print("\n--- INVALID AGES ---")
display(df[(df["Age"] < 0) | (df["Age"] > 100)])

print("\n--- NEGATIVE FARES ---")
display(df[df["Fare"] < 0])

# Detect unusually high fares using the IQR method
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

print("\n--- HIGH FARE OUTLIERS ---")
print("Upper fare limit:", round(upper_limit, 2))
print("Number of high-fare records:", (df["Fare"] > upper_limit).sum())
```

```python

import os
import shutil
from google.colab import files
import matplotlib.pyplot as plt

# Create a folder for the charts
os.makedirs("eda_charts", exist_ok=True)

# Chart 1: Survival count
counts = df["Survived"].value_counts().reindex([0, 1], fill_value=0)

plt.figure(figsize=(7, 4))
plt.bar(["Did not survive", "Survived"], counts.values)
plt.title("Titanic Passenger Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.savefig("eda_charts/survival_count.png")
plt.close()

# Chart 2: Age distribution
plt.figure(figsize=(7, 4))
plt.hist(df["Age"].dropna(), bins=20, edgecolor="black")
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.savefig("eda_charts/age_distribution.png")
plt.close()

# Chart 3: Average fare by survival
avg_fare = df.groupby("Survived")["Fare"].mean().reindex([0, 1])

plt.figure(figsize=(7, 4))
plt.bar(["Did not survive", "Survived"], avg_fare.values)
plt.title("Average Fare by Survival Status")
plt.xlabel("Survival Status")
plt.ylabel("Average Fare")
plt.tight_layout()
plt.savefig("eda_charts/average_fare.png")
plt.close()

# Chart 4: Survival rate by sex
survival_by_sex = df.groupby("Sex_male")["Survived"].mean() * 100

labels = [
    "Female" if value == 0 else "Male"
    for value in survival_by_sex.index
]

plt.figure(figsize=(7, 4))
plt.bar(labels, survival_by_sex.values)
plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)
plt.tight_layout()
plt.savefig("eda_charts/survival_by_sex.png")
plt.close()

# Create a ZIP file containing all charts
shutil.make_archive("eda_charts", "zip", "eda_charts")

print("Charts saved successfully!")
print(os.listdir("eda_charts"))

# Download the ZIP file
files.download("eda_charts.zip")
```
